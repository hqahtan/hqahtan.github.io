## 5. Connecting to an SQLite Database

The `sqlite3` library is included with Python and is available in Google Colab.


### Creating a connection to the database


```python
def create_connection(db_file):
    """Create a connection to the SQLite database specified by db_file."""
    conn = None
    try:
        conn = sqlite3.connect(db_file)
    except Error as e:
        print(e)
    return conn
```

### Querying the database


```python
def run_query(conn, query):
    """Execute a SQL query and return all result rows."""
    cur = conn.cursor()
    cur.execute(query)
    results = cur.fetchall()
    return results
```

### Creating a DataFrame from a database table


```python
def convert_db_table_to_DF(conn, table):
    # Get attribute names
    header_query = "SELECT name FROM pragma_table_info('" + table + "') ORDER BY cid;"
    cols_init = run_query(conn, header_query)
    cols = [cols_init[i][0] for i in range(len(cols_init))]

    # Get table records
    content_query = "SELECT * FROM " + table
    data = run_query(conn, content_query)

    return pd.DataFrame(data, columns=cols)
```

### Example: list the tables in the football database


```python
database = "sample_data/football.db"
conn = create_connection(database)

with conn:
    query = 'SELECT name FROM sqlite_schema WHERE type = "table"'
    data = run_query(conn, query)
    print(data)
```
